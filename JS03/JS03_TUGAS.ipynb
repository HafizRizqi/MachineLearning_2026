1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan

```python
# Menghapus kolom yang tidak digunakan
df = df.drop(columns=['id', 'Unnamed: 32'])

print("Jumlah variabel yang dapat digunakan:", df.shape[1])

print("\nVariabel yang dapat digunakan:")
print(df.columns.tolist())
```

2. Lakukan proses encoding pada kolom "diagnosis"

```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['diagnosis'] = le.fit_transform(df['diagnosis']) # proses encoding
print(df['diagnosis'].unique())
```

```python
df.head()
```

3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
# Mengambil kolom numerik selain diagnosis
kolom_numerik = df.drop(columns=['diagnosis']).select_dtypes(
    include=['int64', 'float64']
).columns

# Membuat objek StandardScaler
scaler = StandardScaler()

# Melakukan standardisasi hanya pada fitur
df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])

print("Kolom yang distandardisasi:")
print(kolom_numerik.tolist())

print("\nHasil standardisasi:")
print(df.head())
```

4. Lakukan proses seleksi fitur. Anda dapat menggunakan SelectKBest.

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

# Memisahkan fitur dan target
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

# Membuat pipeline preprocessing
preprocess = Pipeline([
    ("scaler", StandardScaler())
])

# Feature selection
selector_filter = SelectKBest(score_func=f_classif, k=5)

# Pipeline
pipe_filter = Pipeline([
    ("prep", preprocess),
    ("sel", selector_filter),
    ("clf", LogisticRegression(max_iter=1000))
])
```

```python
# Seleksi fitur dengan SelectKBest
selector_filter = SelectKBest(score_func=f_classif, k=5)

# Membuat pipeline seleksi fitur dan Logistic Regression
pipe_filter = Pipeline([
    ("prep", preprocess),
    ("sel", selector_filter),
    ("clf", LogisticRegression(max_iter=1000))
])
```

5. Lakukan proses pengujian dengan model Logistic Regression seperti pada praktikum 1.

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    stratify=y,
    test_size=0.2,
    random_state=42
)

pipe_filter.fit(X_train, y_train)

pred = pipe_filter.predict(X_test)

print("=== Filter (ANOVA) + LR ===")
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
# 1) Nama fitur setelah preprocess
feat_names = pipe_filter.named_steps["prep"].get_feature_names_out()
print("Nama fitur:", feat_names)

# 2) Mask & skor fitur terpilih (SelectKBest)
sel = pipe_filter.named_steps["sel"]

mask = sel.get_support()
selected_names = feat_names[mask]
selected_scores = sel.scores_[mask]

top = sorted(
    zip(selected_names, selected_scores),
    key=lambda t: t[1],
    reverse=True
)[:10]

print("Top fitur:", top)
```
