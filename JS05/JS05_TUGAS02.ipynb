```python
import pandas as pd

# Membaca dataset
data = pd.read_csv('spam.csv', encoding='latin1')

# Mengambil kolom label dan pesan
data = data.iloc[:, :2]

# Memberikan nama kolom
data.columns = ['label', 'message']

# Melihat data
print(data.head())
print(data.shape)
```

Buat model klasfikasi Multinomial Naive Bayes dengan Ketentuan pertama

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# MEMISAHKAN FITUR DAN LABEL
X = data['message']
y = data['label']

# TRAIN TEST SPLIT
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=50
)

# 1. COUNT VECTORIZER
bow = CountVectorizer(
    stop_words='english'
)

X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

# Model
mnb_bow = MultinomialNB()

mnb_bow.fit(
    X_train_bow,
    y_train
)

# Prediksi
y_train_pred_bow = mnb_bow.predict(X_train_bow)
y_test_pred_bow = mnb_bow.predict(X_test_bow)

# Evaluasi
acc_train_bow = accuracy_score(
    y_train,
    y_train_pred_bow
)

acc_test_bow = accuracy_score(
    y_test,
    y_test_pred_bow
)

print("\n=== COUNT VECTORIZER ===")
print("Akurasi Train:", acc_train_bow)
print("Akurasi Test :", acc_test_bow)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred_bow))

print("\nClassification Report:")
print(classification_report(y_test, y_test_pred_bow))
```

Buat model klasfikasi Multinomial Naive Bayes dengan Ketentuan Kedua

```python
# 2. TF-IDF
tfidf = TfidfVectorizer(
    stop_words='english'
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# Model
mnb_tfidf = MultinomialNB()

mnb_tfidf.fit(
    X_train_tfidf,
    y_train
)

# Prediksi
y_train_pred_tfidf = mnb_tfidf.predict(
    X_train_tfidf
)

y_test_pred_tfidf = mnb_tfidf.predict(
    X_test_tfidf
)

# Evaluasi
acc_train_tfidf = accuracy_score(
    y_train,
    y_train_pred_tfidf
)

acc_test_tfidf = accuracy_score(
    y_test,
    y_test_pred_tfidf
)

print("\n=== TF-IDF ===")
print("Akurasi Train:", acc_train_tfidf)
print("Akurasi Test :", acc_test_tfidf)

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test,
    y_test_pred_tfidf
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_test_pred_tfidf
))

# PERBANDINGAN
print("\n=== PERBANDINGAN ===")

print(
    f"CountVectorizer : "
    f"{acc_test_bow * 100:.2f}%"
)

print(
    f"TF-IDF          : "
    f"{acc_test_tfidf * 100:.2f}%"
)
```

CountVectorizer

CountVectorizer merepresentasikan teks berdasarkan jumlah kemunculan kata dalam dokumen.
Pada dataset spam, pola frekuensi kata tertentu dapat menjadi informasi yang cukup kuat untuk membedakan pesan spam dan ham. Misalnya, pesan spam sering memiliki kata-kata yang berkaitan dengan hadiah, penawaran, kontes, atau promosi.
Pada percobaan ini, CountVectorizer menghasilkan:
Akurasi testing = 98,30%

TF-IDF

TF-IDF memberikan bobot berdasarkan tingkat kepentingan suatu kata terhadap dokumen. Kata yang terlalu sering muncul di banyak dokumen akan mendapatkan bobot yang lebih rendah.
Pada dataset ini, representasi tersebut justru menghasilkan:
Akurasi testing = 96,05%

Sehingga pada percobaan ini, TF-IDF menghasilkan akurasi lebih rendah dibandingkan CountVectorizer.
